```python
import numpy as np
```

```python
def compute_output_size_1d(input_array, kernel_array):
    input_length = len(input_array)
    kernel_length = len(kernel_array)
    conv_length = input_length - kernel_length + 1
    return conv_length

# -----------------------------------------------
# Example:
input_array = np.array([1, 2, 3, 4, 5])
kernel_array = np.array([1, 0, -1])
print(compute_output_size_1d(input_array, kernel_array))
```

```python
# Your code here:
# -----------------------------------------------

def convolve_1d(input_array, kernel_array):
    # Tip: start by initializing an empty output array (you can use your function above to calculate the correct size).
    # Then fill the cells in the array with a loop.
    length = compute_output_size_1d(input_array, kernel_array)
    convolution = [None] * length
    for i in range(length):
        vec_input = input_array[list(range(i, i + length))]
        dot_product = np.dot(vec_input, kernel_array)
        convolution[i] = dot_product
    return convolution

# -----------------------------------------------
# Another tip: write test cases like this, so you can easily test your function.
input_array = np.array([1, 2, 3, 4, 5])
kernel_array = np.array([1, 0, -1])
print(convolve_1d(input_array, kernel_array))
```

```python
# Your code here:
# -----------------------------------------------
def compute_output_size_2d(input_matrix, kernel_matrix):
    input_height = np.size(input_matrix, 0)
    input_width = np.size(input_matrix, 1)
    kernel_height = np.size(kernel_matrix, 0)
    kernel_width = np.size(kernel_matrix, 1)
    dimensions = (input_height - kernel_height + 1, input_width - kernel_width + 1)
    return dimensions
# -----------------------------------------------
# Example:
input_array = np.array([[1, 2, 3, 4, 5],
                       [5, 4, 3, 2, 1],
                       [9, 8, 7, 6, 5]])
kernel_array = np.array([[1, 0, -1],
                        [1, 0, -1]])
print(compute_output_size_2d(input_array, kernel_array))
```

```python
def convolute_2d(input_matrix, kernel_matrix):
    # Tip: same tips as above, but you might need a nested loop here in order to
    # define which parts of the input matrix need to be multiplied with the kernel matrix.
    size = compute_output_size_2d(input_matrix, kernel_matrix)
    convolution = np.empty(size)
    height_conv = size[0]
    width_conv = size[1]
    height_kernel = np.size(kernel_matrix, 0)
    width_kernel = np.size(kernel_matrix, 1)
    for col in range(width_conv):
        for row in range(height_conv):
            matrix_part = input_matrix[row:(row + height_kernel), col:(col + width_kernel)]
            matrix_part_height = np.size(matrix_part, 0)
            dot_products_row = [0] * matrix_part_height
            for row_part in range(matrix_part_height):
                dot_products_row[row_part] = np.dot(matrix_part[row_part], kernel_matrix[row_part])
            dot_product = sum(dot_products_row)
            convolution[row, col] = dot_product
    return convolution
# -----------------------------------------------
# Example:
input_matrix = np.array([[1, 2, 3, 4, 5],
                         [5, 4, 3, 2, 1],
                         [5, 8, 1, 4, 6]])
kernel_matrix = np.array([[1, 0, -1],
                          [1, 0, -1]])
convolute_2d(input_matrix, kernel_matrix)
```
